## 1. The mass of the atmosphere
We calculate the mass of the atmosphere as stated in *Bohren - Atmospheric Thermodynamics*, p.171. We start by calculating the mass per square meter according to the simple formula
$$M = \int_0^{\infty} \rho A dz$$
where A is the Earth's surface area and $\rho$ the air density. Using the hydrostatic equation $dp = -\rho g dz$ we have
$$M = - \frac{A}{g}\int_{p_0}^0 dp = \frac{Ap_0}{g}$$
where $p_0$ is the pressure at ground level.

```python
p_0 = 10 ** 5 * ureg.pascal
p_0_bu = p_0.to_base_units()
```

The radius of the Earth is 6400 km

```python
R = 6400 * 1000 * ureg.m
```

so its surface area is

```python
A = 4 * 3.14 * R ** 2
```

We define the gravity accelleration close to the Earth's surface

```python
g = 9.8 * ureg.m / ureg.s ** 2
```

Finally the mass of the Earth in Kg is

```python
M = A * p_0_bu / g
M 
```

That is 5 Millions Gigatons. We can compare that number with the amount of $CO_2$ that is released every year in the atmosphere, around 40 Gigatons

```python
M_CO2 = 40 * 10 ** 9 * 1000 * ureg.kilogram
M_CO2
```

```python
M_CO2 / M
```

## 2. Precipitable water in the atmosphere
Here we want to know what would be the depth $d_w$ of the liquid water if the vapor in the atmosphere were to condense and fall on the surface of a flat Earth. Since the water mass is conserved we have
$$\rho_w d_w = \int_0^{\infty} \rho_v dz$$
where $\rho_w$ is the water density and $\rho_v$ is the water vapor density. Assuming the water vapor being an ideal gas the vapor pressure $e$ follows the same law
$$e = \rho_v R_v T $$
where $R_v$ is the water vapor gas constant. The water depth is
$$d_w = \frac{1}{\rho_w R_v} \int_0^{\infty} \frac{e}{T} dz$$
We consider the water vapor pressure at each height z to be at saturation and represented by the Clausius-Clapeyron equation. From the chain rule we can write
$$\frac{de}{dz} = \frac{de}{dT} \frac{dT}{dz} = -\Gamma \frac{de}{dT}$$
The Clausius-Clapeyron equation can be written as
$$\frac{de}{dT} = \frac{L_v e}{R_v T^2}$$
so that 
$$\frac{de}{dz} = -\Gamma \frac{L_v e}{R_v T^2}$$
We can substitute this expression in the integral to use the pressure instead of the heigth for the integration
$$d_w = -\frac{1}{\rho_w \Gamma L_v} \int_{e_0}^0 Tde $$
If we assume the temperature T to be constant at $T_0$ we have
$$d_w = \frac{T_0 e_0}{\rho_w L_v \Gamma} $$

As in Bohren et al., we assume some standard values for the variables in this last equation to have a rough value for the depth of the precipitabale water in the atmosphere.

```python
t_0 = 25.0 * ureg.degree_Celsius
T_0 = t_0.to('degK')
T_0
```

```python
e_0 = 3200 * ureg.pascal
e_0_bu = e_0.to_base_units()
e_0_bu
```

```python
Gamma = 6.5 * ureg.degree_Kelvin / ureg.kilometer
Gamma_bu = Gamma.to_base_units()
Gamma_bu
```

```python
r_w = 998.21 * ureg.kilogram / ureg.m ** 3
r_w_bu = r_w.to_base_units()
r_w_bu
```

```python
L_v = 2.501 * 10 ** 6 * ureg.joule / ureg.kilogram
L_v_bu = L_v.to_base_units()
L_v_bu
```

The value for the depth is close to that calculated in the book, about 6 cm.

```python
d_w = T_0 * e_0_bu / (r_w_bu * L_v_bu * Gamma_bu)
d_w.to('cm') 
```

For a detailed description see Bohren et al. par. 6.1.
